```python
def assign_pop_bin(pop):
    if pop <= 100:
        return '25 - 100'
    elif pop <= 500:
        return '101 - 500'
    elif pop <= 1000:
        return '501 - 1,000'
    elif pop <= 3300:
        return '1,001 - 3,300'
    elif pop <= 10000:
        return '3,301 - 10,000'
    elif pop <= 50000:
        return '10,001 - 50,000'
    elif pop <= 100000:
        return '50,001 - 100,000'
    elif pop <= 1000000:
        return '100,001 - 1,000,000'
    else:
        return '> 1,000,000'
```

```python
active_sdwis['Population_bin'] = active_sdwis['POPULATION_SERVED_COUNT'].apply(assign_pop_bin)
```

```python
# Total count of active Community Water Systems in the US, aligned to population bins for scaling 
bin_order = [
    '25 - 100', '101 - 500', '501 - 1,000', '1,001 - 3,300', 
    '3,301 - 10,000', '10,001 - 50,000', '50,001 - 100,000', 
    '100,001 - 1,000,000', '> 1,000,000'
]

national_pws_vector = active_sdwis['Population_bin'].value_counts().reindex(bin_order)

national_pws_vector
```

Established the National Universe by pulling the total count of all active Community Water Systems from the SDWIS database and organizing them into specific population bins.

```python
ep_table = pd.read_csv('ep_table.csv')
```

```python
ep_table.head()
```

```python
ep_table.info()
```

```python
ep_table['Ep_Population'] = ep_table['Ep_Population'].round()
```

```python
ep_table.head()
```

```python
# Count of all unique Entry Points (EPs) by Population Bin. Basically, a total sample size to calculate the risk of contamination
total_eps_per_bin = ep_table.groupby('Population_bin')['FacilityID'].nunique()
total_eps_per_bin
```

```python
ep_table['Total_PFAS'] =  ep_table['PFOA'] + ep_table['PFOS']
```

```python
thresholds = [4, 8, 12, 20, 40]
```

```python
impacted_df = pd.DataFrame({'Population_bin': bin_order})
```

```python
# for each regulatory threshold:
# Numerator: unique Eps that exceed the threshold
# Denominator: All unique EPs in that Population bin (total_eps_per_bin)
# Result: The Percent of impact for each bin 
for t in thresholds:
    threshold_ug = t / 1000
    
    # Matching  Cost Notebook 'OR' logic
    exceeding = (ep_table['PFOA'] > threshold_ug) | (ep_table['PFOS'] > threshold_ug)
    
    impacted_counts = ep_table[exceeding].groupby('Population_bin')['FacilityID'].nunique().reindex(bin_order)
    
    impacted_df[f'impacted_percent_{t}ng'] = (impacted_counts.values / total_eps_per_bin.reindex(bin_order).values).round(4) * 100
    
impacted_df = impacted_df.fillna(0)
impacted_df
```

Developed the Impacted Dataframe using the UCMR5 sample data, where we calculated the specific percentage of unique entry points that exceeded each regulatory threshold based on a logical "OR" comparison of PFOA and PFOS levels

```python
unique_eps = ep_table.drop_duplicates(subset=['PWSID', 'FacilityID'])

total_eps = len(unique_eps)
total_pws = ep_table['PWSID'].nunique()

M = total_eps / total_pws

print(f"Total Unique EPs: {total_eps}")
print(f"Total Unique PWS: {total_pws}")
print(f"National Multiplier: {M:.4f}")
```

Calculated a National Multiplier representing the average number of entry points per water system.

```python
final_national_impact = pd.DataFrame({'Population_bin': bin_order})
final_national_impact['National_PWS_Count'] = national_universe['National_Systems'].values

for t in thresholds:
    p_col = f'impacted_percent_{t}ng'
    
    # Formula: N (National Systems) * M (EPs per System) * P (Percent Impacted / 100)
    # We divide by 100 because the table currently has values like 11.98 instead of 0.1198
    final_national_impact[f'Estimated_EPs_to_Treat_{t}ng'] = (
        final_national_impact['National_PWS_Count'] * M * (impacted_df[p_col] / 100)
    ).round(0)

final_national_impact
```

Finally, generated the Final Scaling Table by multiplying the national system counts by this multiplier and the bin-specific impact percentages
